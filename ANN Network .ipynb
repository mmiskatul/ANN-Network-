### Importing the library

```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd 
```

```python
dataset =pd.read_csv('customer_data.csv')
dataset.head()
```

```python
x =dataset.iloc[:,3:13].values 
y =dataset.iloc[:,13].values 
```

```python
x,y
```

```python
#Encoding categorical data 
from sklearn.preprocessing import LabelEncoder,OneHotEncoder 
labelencoder_x_1 =LabelEncoder() 
x[:,1] =labelencoder_x_1.fit_transform(x[:,1]) 
labelencoder_x_2 =LabelEncoder() 
x[:,2] =labelencoder_x_2.fit_transform(x[:,2]) 
x
```

```python
# Split the dataset into the train and test data set 

from sklearn.model_selection import train_test_split 
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.2,random_state=0) 
```

```python
len(dataset)
```

```python
len(x_train)
```

## Feature Scaling 

```python
## Standrization and Normalization 
from sklearn.preprocessing import StandardScaler 
sc = StandardScaler() 
x_train_sc =sc.fit_transform(x_train) 
x_test_sc =sc.transform(x_test) 
```

```python
# Classical ML 
from sklearn.tree import DecisionTreeClassifier 
model_dt =DecisionTreeClassifier() 
model_dt.fit(x_train_sc,y_train) 
pred =model_dt.predict(x_test_sc) 
```

```python
#making the confusion metrix 
from sklearn.metrics import confusion_matrix 
cm =confusion_matrix(y_test,pred)  
print(cm)
```

```python
from sklearn.metrics import accuracy_score 

print(round(accuracy_score(y_test,pred),2)*100)
```

```python
# Classical ML 
from sklearn.ensemble import RandomForestClassifier 
model_rf =RandomForestClassifier() 
model_rf.fit(x_train_sc,y_train) 
pred2 =model_rf.predict(x_test_sc) 
```

```python
cm =confusion_matrix(y_test,pred2)   
print(cm)
```

```python
print(round(accuracy_score(y_test,pred2),2)*100)
```

## DEEP LEARNING 

```python
#Import the library
import keras 
from keras.models import Sequential 
from keras.layers import Dense ,Input
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

import tensorflow as tf
```

```python
# initilissing the ANN 
classifier =Sequential() 

# Add the inpput layer and first hidden layer  
classifier.add(Input(shape=(10,)))
classifier.add(Dense(units=6, activation='relu', kernel_initializer='uniform'))

# Second hidden layer
classifier.add(Dense(units=6, activation='relu', kernel_initializer='uniform'))

# Output layer 
classifier.add(Dense(units=1, activation='sigmoid', kernel_initializer='uniform'))
```

```python
# compling the ANN 
classifier.compile(optimizer= 'adam' ,loss = 'binary_crossentropy', metrics=['accuracy']) 
y_train[0] 

# Fiting the ANN to training set 
classifier.fit(x_train_sc,y_train,batch_size=10,epochs=200)
```

```python
y_pred_ann =classifier.predict(x_test_sc) 
y_pred_ann =(y_pred_ann>0.5) 

cm3 =confusion_matrix(y_test,y_pred_ann) 
print(cm3)
```

```python
from sklearn.metrics import accuracy_score

print(round(accuracy_score(y_test,y_pred_ann),2)*100)
```
